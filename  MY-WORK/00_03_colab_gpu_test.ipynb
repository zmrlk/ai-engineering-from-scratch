```python
import torch

print("PyTorch:", torch.__version__)
print("CUDA dostępna:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
    print(
        "Pamięć GPU:",
        round(torch.cuda.get_device_properties(0).total_memory / 1e9, 1),
        "GB",
    )
```

```python
import time
import torch

size = 2000

a_cpu = torch.randn(size, size)
b_cpu = torch.randn(size, size)

start_cpu = time.perf_counter()
c_cpu = a_cpu @ b_cpu
cpu_time = time.perf_counter() - start_cpu

a_cuda = a_cpu.to("cuda")
b_cuda = b_cpu.to("cuda")

# Rozgrzewka GPU
_ = a_cuda @ b_cuda
torch.cuda.synchronize()

start_cuda = time.perf_counter()
c_cuda = a_cuda @ b_cuda
torch.cuda.synchronize()
cuda_time = time.perf_counter() - start_cuda

print(f"CPU Colaba: {cpu_time:.4f} s")
print(f"NVIDIA T4: {cuda_time:.4f} s")
print(f"Przyspieszenie: {cpu_time / cuda_time:.2f}x")
print("Wynik CUDA znajduje się na:", c_cuda.device)
```
